```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Create dataset
data = {
    "news": [
        "Government announced a new education policy",
        "University released examination results",
        "Scientists published a research paper",
        "Railway department announced new trains",
        "Government opened a new hospital",
        "School announced examination dates",
        "Election commission released polling schedule",
        "Weather department issued a rain warning",
        "Company published its financial report",
        "Court released an official order",
        "Health department published a health report",
        "University announced admission dates",
        "New public library opened in the city",
        "Government announced scholarship applications",
        "Research team presented its scientific findings",
        "Bank announced new customer service timings",

        "Magic plant makes humans live forever",
        "Everyone gets free money by sharing this message",
        "The moon will disappear tomorrow",
        "A secret drink cures all diseases",
        "A magic button gives unlimited internet",
        "Aliens are controlling every mobile phone",
        "A secret machine makes people invisible",
        "Drinking special water gives superpowers",
        "Share this message to become a millionaire",
        "The sun will stop shining next week",
        "A miracle food guarantees eternal life",
        "Every person will receive free gold tomorrow",
        "A secret formula makes cars fly",
        "An unknown website promises instant wealth",
        "A magic tablet cures every illness instantly",
        "Humans can breathe underwater without equipment"
    ],

    "label": [
        "REAL", "REAL", "REAL", "REAL",
        "REAL", "REAL", "REAL", "REAL",
        "REAL", "REAL", "REAL", "REAL",
        "REAL", "REAL", "REAL", "REAL",

        "FAKE", "FAKE", "FAKE", "FAKE",
        "FAKE", "FAKE", "FAKE", "FAKE",
        "FAKE", "FAKE", "FAKE", "FAKE",
        "FAKE", "FAKE", "FAKE", "FAKE"
    ]
}

# Step 2: Convert data into DataFrame
df = pd.DataFrame(data)

# Step 3: Separate input and output
X = df["news"]
y = df["label"]

# Step 4: Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Step 5: Create machine learning pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        stop_words="english",
        ngram_range=(1, 2)
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Step 6: Train the model
model.fit(X_train, y_train)

print("Model training completed successfully!")

# Step 7: Test the model
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("\nModel Evaluation")
print("----------------")
print("Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))

# Step 8: Predict new news
print("\n================================")
print("     FAKE NEWS DETECTION SYSTEM")
print("================================")

while True:

    news = input("\nEnter News: ").strip()

    if news.lower() == "exit":
        print("Thank you for using the system!")
        break

    if not news:
        print("Please enter valid news.")
        continue

    prediction = model.predict([news])[0]

    print("\nPrediction Result:")

    if prediction == "FAKE":
        print("This news is classified as FAKE.")
    else:
        print("This news is classified as REAL.")

    print("\nNote: This is only a machine learning prediction.")
    print("It does not verify whether the news is actually true.")
```
